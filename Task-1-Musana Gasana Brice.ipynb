```python
import pandas as pd
df = pd.read_excel('Dataset for Data Analytics.xlsx')
print(df.head())
```

```python
#checking whether there are null values and their sum
df.isnull().sum().sum()
```

```python
#identifying columns with null values
df.columns[df.isnull().any()]
```

```python
#filling the missing values in the CouponCode column
df = df.fillna('NaN')
```

```python
#checking whether there are no more missing values
df.isnull().sum().sum()
```

```python
#finding duplicates
print("Duplicate rows:", df.duplicated().sum())
```

```python
#checking whether the data types are correct
df.dtypes
```

```python
#saving the cleaned dataset
df.to_excel('Clean Dataset for Data Analytics.xlsx', index=False)
```
